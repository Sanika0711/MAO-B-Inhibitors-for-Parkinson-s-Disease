```python
import torch

from torch_geometric.data import Data
from torch_geometric.loader import DataLoader

from rdkit import Chem
```

```python
def atom_features(atom):
    return [
        atom.GetAtomicNum(),
        atom.GetDegree(),
        atom.GetFormalCharge(),
        int(atom.GetIsAromatic())
    ]
```

```python
def mol_to_graph(mol, y=None):

    node_features = []

    for atom in mol.GetAtoms():
        node_features.append(
            atom_features(atom)
        )

    x = torch.tensor(
        node_features,
        dtype=torch.float
    )

    edge_index = []

    for bond in mol.GetBonds():

        start = bond.GetBeginAtomIdx()
        end = bond.GetEndAtomIdx()

        edge_index.append([start, end])
        edge_index.append([end, start])

    edge_index = torch.tensor(
        edge_index,
        dtype=torch.long
    ).t().contiguous()

    if y is not None:
        y = torch.tensor(
            [y],
            dtype=torch.float
        )

    return Data(
        x=x,
        edge_index=edge_index,
        y=y
    )
```

```python
import torch
import torch.nn as nn

from torch_geometric.nn import (
    GCNConv,
    global_mean_pool
)
```

```python
class GCN(torch.nn.Module):

    def __init__(self):

        super().__init__()

        self.conv1 = GCNConv(4, 64)

        self.conv2 = GCNConv(64, 64)

        self.fc1 = nn.Linear(
            64,
            32
        )

        self.fc2 = nn.Linear(
            32,
            1
        )

    def forward(
        self,
        x,
        edge_index,
        batch
    ):

        x = self.conv1(
            x,
            edge_index
        )

        x = torch.relu(x)

        x = self.conv2(
            x,
            edge_index
        )

        x = torch.relu(x)

        x = global_mean_pool(
            x,
            batch
        )

        x = torch.relu(
            self.fc1(x)
        )

        x = self.fc2(x)

        return x
```

```python
from rdkit import Chem
from torch_geometric.data import Data
import torch

def mol_to_graph(mol, y=None):

    atom_features = []

    for atom in mol.GetAtoms():

        atom_features.append([
            atom.GetAtomicNum(),
            atom.GetDegree(),
            atom.GetFormalCharge(),
            atom.GetTotalNumHs(),
            int(atom.GetIsAromatic()),
            atom.GetHybridization().real
        ])

    x = torch.tensor(
        atom_features,
        dtype=torch.float
    )

    edge_index = []

    for bond in mol.GetBonds():

        i = bond.GetBeginAtomIdx()
        j = bond.GetEndAtomIdx()

        edge_index.append([i, j])
        edge_index.append([j, i])

    edge_index = torch.tensor(
        edge_index,
        dtype=torch.long
    ).t().contiguous()

    if y is not None:

        y = torch.tensor(
            [y],
            dtype=torch.float
        )

        return Data(
            x=x,
            edge_index=edge_index,
            y=y
        )

    return Data(
        x=x,
        edge_index=edge_index
    )
```

```python
import torch
import torch.nn.functional as F

from torch.nn import Linear
from torch_geometric.nn import GCNConv
from torch_geometric.nn import global_mean_pool
```

```python
class GCN(torch.nn.Module):

    def __init__(self):

        super().__init__()

        self.conv1 = GCNConv(6, 64)
        self.conv2 = GCNConv(64, 64)

        self.fc1 = Linear(64, 32)
        self.fc2 = Linear(32, 1)

    def forward(self, x, edge_index, batch):

        x = self.conv1(x, edge_index)
        x = F.relu(x)

        x = self.conv2(x, edge_index)
        x = F.relu(x)

        x = global_mean_pool(x, batch)

        x = F.relu(self.fc1(x))

        return self.fc2(x)
```

```python
import torch
import torch.nn.functional as F

from torch.nn import Linear
from torch_geometric.nn import GCNConv
from torch_geometric.nn import global_mean_pool

class BetterGCN(torch.nn.Module):

    def __init__(self):

        super().__init__()

        self.conv1 = GCNConv(9,128)

        self.conv2 = GCNConv(128,128)

        self.fc1 = Linear(128,64)

        self.fc2 = Linear(64,1)

    def forward(
        self,
        x,
        edge_index,
        batch
    ):

        x = self.conv1(
            x,
            edge_index
        )

        x = F.relu(x)

        x = F.dropout(
            x,
            p=0.2,
            training=self.training
        )

        x = self.conv2(
            x,
            edge_index
        )

        x = F.relu(x)

        x = global_mean_pool(
            x,
            batch
        )

        x = F.relu(
            self.fc1(x)
        )

        return self.fc2(x)
```

```python
# ==== Applicability Domain check (fully self-contained version) ====
# This version does NOT depend on any earlier cells in your notebook having
# been run. It reloads your frozen dataset and rebuilds everything from
# scratch, so you can run it in a brand-new kernel session on its own.

import pandas as pd
import numpy as np
import random
from rdkit import Chem, DataStructs
from rdkit.Chem import rdFingerprintGenerator

# ---- Step 0: Load your frozen CLEANED dataset ----
# IMPORTANT: replace this filename with your actual frozen cleaned CSV
# (the one saved by your "chembl_maob_clean_<date>.csv" cell, NOT the raw one).
CLEAN_CSV_PATH = "chembl_maob_clean_2026-06-20.csv"  # <-- change this to your real filename

df3 = pd.read_csv(CLEAN_CSV_PATH)
df3['Mol'] = df3['canonical_smiles'].apply(Chem.MolFromSmiles)
df3 = df3[df3['Mol'].notna()].reset_index(drop=True)  # drop any that failed to parse
print(f"Loaded frozen dataset: {df3.shape[0]} compounds")

# ---- Step 1: Rebuild the same Morgan fingerprint generator you used for the RF model ----
morgan_gen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=2048)

train_fps = [morgan_gen.GetFingerprint(mol) for mol in df3['Mol']]
print(f"Training set fingerprints built: {len(train_fps)}")

# ---- Step 2: Your nine designed analogues (SMILES hard-coded here, no dependency on notebook state) ----
query_compounds = {
    "Lead":        "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(C)c1",
    "m-F":         "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(F)c1",
    "m-Cl":        "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(Cl)c1",
    "m-Br":        "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(Br)c1",
    "m-CF3":       "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(C(F)(F)F)c1",
    "m-OCH3":      "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(OC)c1",
    "3,4-diOCH3":  "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cc(OC)c(OC)cc1",
    "Pyridyl":     "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccnc1",
    "N-Propargyl": "C#CCN1N=C(c2ccc(O)cc2)CC1c1cccc(C)c1",
}

# ---- Step 3: Max Tanimoto similarity of each query compound to the training set ----
results = []
for name, smi in query_compounds.items():
    mol = Chem.MolFromSmiles(smi)
    fp = morgan_gen.GetFingerprint(mol)
    sims = DataStructs.BulkTanimotoSimilarity(fp, train_fps)
    results.append({"Compound": name, "Max_Tanimoto_to_training_set": round(max(sims), 3)})

ad_df = pd.DataFrame(results)
print(ad_df)

# ---- Step 4: Data-driven threshold from the training set's OWN internal similarity ----
sample_idx = random.sample(range(len(train_fps)), min(500, len(train_fps)))
nn_sims = []
for i in sample_idx:
    others = train_fps[:i] + train_fps[i+1:]
    sims = DataStructs.BulkTanimotoSimilarity(train_fps[i], others)
    nn_sims.append(max(sims))

threshold = np.percentile(nn_sims, 5)
print(f"\nTraining set nearest-neighbor similarity: mean={np.mean(nn_sims):.3f}, "
      f"5th percentile={threshold:.3f}")

ad_df["Outside_AD"] = ad_df["Max_Tanimoto_to_training_set"] < threshold
print(f"\nFinal applicability domain assessment (threshold = {threshold:.3f}):")
print(ad_df)
```

```python
# ==== Check: are Lead and m-Cl literal members of the training set? ====
# The Tanimoto check found Lead and m-Cl both at 1.000 similarity to their
# nearest training-set neighbour. 1.000 similarity on a 2048-bit Morgan
# fingerprint essentially always means either (a) the exact same molecule,
# or (b) two different molecules that happen to produce identical fingerprints
# (rare, but possible for very small/simple structures). This cell checks
# which one it actually is.

from rdkit import Chem

# ---- Load your frozen dataset (adjust filename to match your actual saved CSV) ----
import pandas as pd
CLEAN_CSV_PATH = "chembl_maob_clean_2026-06-20.csv"  # <-- change to your real filename
df3 = pd.read_csv(CLEAN_CSV_PATH)

lead = "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(C)c1"
mCl  = "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(Cl)c1"

def check_exact_match(name, query_smiles, df, smiles_col="canonical_smiles"):
    query_mol = Chem.MolFromSmiles(query_smiles)
    query_canonical = Chem.MolToSmiles(query_mol)  # canonicalize for fair comparison

    matches = []
    for idx, row in df.iterrows():
        train_mol = Chem.MolFromSmiles(row[smiles_col])
        if train_mol is None:
            continue
        train_canonical = Chem.MolToSmiles(train_mol)
        if train_canonical == query_canonical:
            matches.append(row)

    print(f"\n--- {name} ---")
    print(f"Query canonical SMILES: {query_canonical}")
    if matches:
        print(f"EXACT MATCH found: {len(matches)} identical compound(s) in training set.")
        for m in matches:
            # print whatever identifying/activity columns you have — adjust column
            # names below to match your actual dataframe
            print(m[[c for c in ['molecule_chembl_id', 'canonical_smiles', 'pIC50'] if c in df.columns]])
    else:
        print("No exact match found — the 1.000 similarity is a fingerprint collision "
              "between two DIFFERENT molecules, not a duplicate. Worth a closer look "
              "at which training compound it collided with.")
        # Optionally: find and show the actual nearest neighbour for inspection
        from rdkit.Chem import rdFingerprintGenerator
        from rdkit import DataStructs
        morgan_gen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=2048)
        qfp = morgan_gen.GetFingerprint(query_mol)
        best_sim, best_smi = 0, None
        for smi in df[smiles_col]:
            m = Chem.MolFromSmiles(smi)
            if m is None:
                continue
            fp = morgan_gen.GetFingerprint(m)
            sim = DataStructs.TanimotoSimilarity(qfp, fp)
            if sim > best_sim:
                best_sim, best_smi = sim, smi
        print(f"Nearest training compound (similarity={best_sim:.3f}): {best_smi}")

check_exact_match("Lead", lead, df3)
check_exact_match("m-Cl", mCl, df3)
```

```python
# ==== CNS MPO Score Calculation (Wager et al. 2010/2016 methodology) ====
# Computes the six-parameter CNS MPO desirability score for each compound
# and reproduces the Figure 8 bar chart from the manuscript.

from rdkit import Chem
from rdkit.Chem import Descriptors, Crippen, rdMolDescriptors
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import pandas as pd
import numpy as np

# ---- Step 1: Desirability functions (piecewise-linear, per Wager et al.) ----
# Each returns a score between 0 (least desirable) and 1 (most desirable).

def score_monotonic_decreasing(value, t1, t0):
    """Score = 1 below t1, linearly decreases to 0 at t0, 0 beyond."""
    if value <= t1:
        return 1.0
    elif value >= t0:
        return 0.0
    else:
        return (t0 - value) / (t0 - t1)

def score_hump(value, low0, low1, high1, high0):
    """Hump-shaped: 0 -> 1 -> 1 -> 0 across four inflection points (used for TPSA)."""
    if value <= low0 or value >= high0:
        return 0.0
    elif low0 < value < low1:
        return (value - low0) / (low1 - low0)
    elif low1 <= value <= high1:
        return 1.0
    else:  # high1 < value < high0
        return (high0 - value) / (high0 - high1)

def clogp_score(clogp):
    return score_monotonic_decreasing(clogp, t1=3, t0=5)

def clogd_score(clogd):
    return score_monotonic_decreasing(clogd, t1=2, t0=4)

def mw_score(mw):
    return score_monotonic_decreasing(mw, t1=360, t0=500)

def tpsa_score(tpsa):
    return score_hump(tpsa, low0=20, low1=40, high1=90, high0=120)

def hbd_score(hbd):
    return score_monotonic_decreasing(hbd, t1=0.5, t0=3.5)

def pka_score(pka):
    return score_monotonic_decreasing(pka, t1=8, t0=10)


# ---- Step 2: Compute CNS MPO for a single compound ----
def calc_cns_mpo(smiles, assumed_pka=2.5):
    """
    assumed_pka: the pKa of the compound's most basic center. Since RDKit
    has no built-in pKa predictor, this follows the same approach used in
    the manuscript (Section 2.7): LogD is approximated as equal to LogP
    (compounds are predominantly neutral at physiological pH 7.4), and the
    most basic center's pKa is taken from the pyrazoline nitrogen (~2-3).
    Override this per-compound if you have ADMET-predicted pKa values instead.
    """
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        raise ValueError(f"Could not parse SMILES: {smiles}")

    mw = Descriptors.MolWt(mol)
    clogp = Crippen.MolLogP(mol)
    clogd = clogp  # approximation per manuscript methodology (neutral at pH 7.4)
    tpsa = rdMolDescriptors.CalcTPSA(mol)
    hbd = rdMolDescriptors.CalcNumHBD(mol)
    pka = assumed_pka

    scores = {
        "LogP_score": clogp_score(clogp),
        "LogD_score": clogd_score(clogd),
        "MW_score": mw_score(mw),
        "TPSA_score": tpsa_score(tpsa),
        "HBD_score": hbd_score(hbd),
        "pKa_score": pka_score(pka),
    }
    total = sum(scores.values())

    return {
        "MW": round(mw, 2), "ClogP": round(clogp, 2), "ClogD": round(clogd, 2),
        "TPSA": round(tpsa, 1), "HBD": hbd, "pKa": pka,
        **{k: round(v, 3) for k, v in scores.items()},
        "CNS_MPO_Score": round(total, 2),
    }


# ---- Step 3: Run it on your nine designed analogues ----
compounds = {
    "Lead":        "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(C)c1",
    "m-F":         "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(F)c1",
    "m-Cl":        "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(Cl)c1",
    "m-Br":        "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(Br)c1",
    "m-CF3":       "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(C(F)(F)F)c1",
    "m-OCH3":      "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccc(OC)c1",
    "3,4-diOCH3":  "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cc(OC)c(OC)cc1",
    "Pyridyl":     "CC(=O)N1N=C(c2ccc(O)cc2)CC1c1cccnc1",
    "N-Propargyl": "C#CCN1N=C(c2ccc(O)cc2)CC1c1cccc(C)c1",
}

results = []
for name, smi in compounds.items():
    row = calc_cns_mpo(smi)
    row["Compound"] = name
    results.append(row)

mpo_df = pd.DataFrame(results).set_index("Compound")
mpo_df = mpo_df[["MW", "ClogP", "ClogD", "TPSA", "HBD", "pKa",
                  "LogP_score", "LogD_score", "MW_score", "TPSA_score",
                  "HBD_score", "pKa_score", "CNS_MPO_Score"]]
print(mpo_df)


# ---- Step 4: Reproduce Figure 8 (status-coded bar chart) ----
status_map = {
    "Lead": "final", "m-OCH3": "final", "3,4-diOCH3": "final",
    "m-F": "screened", "m-Cl": "screened", "m-Br": "screened", "m-CF3": "screened",
    "Pyridyl": "eliminated", "N-Propargyl": "eliminated",
}
color_map = {"final": "#0F766E", "screened": "#8FD3C7", "eliminated": "#D64545"}
colors = [color_map[status_map[c]] for c in mpo_df.index]

fig, ax = plt.subplots(figsize=(13, 7))
bars = ax.bar(mpo_df.index, mpo_df["CNS_MPO_Score"], color=colors, width=0.65)

for bar, val in zip(bars, mpo_df["CNS_MPO_Score"]):
    ax.text(bar.get_x() + bar.get_width() / 2, val + 0.07, f"{val:.2f}",
             ha="center", va="bottom", fontsize=13, fontweight="bold", color="#222222")

ax.axhline(4.0, color="#333333", linestyle="--", linewidth=1.3)
ax.set_ylim(0, 6.4)
ax.set_ylabel("CNS MPO Score", fontsize=15, fontweight="bold")
ax.set_xlabel("Compound", fontsize=15, fontweight="bold")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

legend_elems = [
    mpatches.Patch(facecolor="#0F766E", label="Final candidate"),
    mpatches.Patch(facecolor="#8FD3C7", label="Screened out (low pIC50)"),
    mpatches.Patch(facecolor="#D64545", label="Eliminated (hepatotoxic)"),
    plt.Line2D([0], [0], color="#333333", linestyle="--", label="CNS MPO threshold (\u22654.0)"),
]
ax.legend(handles=legend_elems, loc="lower right", fontsize=11, frameon=True)

plt.tight_layout()
plt.savefig("cns_mpo_scores.png", dpi=300)
plt.show()
```
